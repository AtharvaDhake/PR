```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.models import Model
from tensorflow.keras.regularizers import l1

data = pd.read_csv("mnist_784_csv.csv").dropna()
X = data.drop("class", axis=1).values

# 3. Normalize
X = X / 255.0

# 4. Build Autoencoder
inp = Input(shape=(784,))

# Encoder with sparsity constraint
encoded = Dense(
    128,
    activation="relu",
    activity_regularizer=l1(0.00001)
)(inp)

# Decoder
decoded = Dense(784, activation="sigmoid")(encoded)

# Create model
autoencoder = Model(inp, decoded)

# 5. Compile
autoencoder.compile(
    optimizer="adam",
    loss="binary_crossentropy"
)

# 6. Train
autoencoder.fit(
    X, X,
    epochs=20,
    batch_size=32,
    validation_split=0.2,
    shuffle=True
)

# 7. Reconstruct images
reconstructed = autoencoder.predict(X[:10])

# 8. Display images
plt.figure(figsize=(12, 4))

for i in range(10):

    # Original image
    plt.subplot(2, 10, i + 1)
    plt.imshow(X[i].reshape(28, 28), cmap="gray")
    plt.axis("off")

    # Reconstructed image
    plt.subplot(2, 10, i + 11)
    plt.imshow(reconstructed[i].reshape(28, 28), cmap="gray")
    plt.axis("off")

plt.show()

print("Sparse Autoencoder implemented successfully!")
```
